```python
def f(x):
    y = 0
    if 0 <= x < 2:
        y = x**2
    elif 2 <= x <= 3:
        y = 4
    elif 3 < x <= 5:
        y = 4 - 2*(x-3)
    return y
```

```python
def shipping_fee(w):
    if w <= 0:
        fee = 0
    elif w <= 1:
        fee = 5
    elif w <= 5:
        fee = 5 + 2*(w-1)
    else:
        fee = min(13 + 1.5*(w-5),20)
    return fee

print(shipping_fee(9))
print(shipping_fee(10))
print(shipping_fee(25))
```

```python
goal = 5000
max_months = 22

balance = 0

# CODE HERE

for i in range(max_months):
    balance = balance * 1.01 + 200
    if balance >= goal:
        print("Goal reached in month", i+1, "!")
        break
    elif i == max_months - 1:
        print("Goal not reached in", max_months, "months")
```

```python
returns = [1.2, -0.5, 35.0, 0.8, -2.1, -50.0, 3.3, 20, -20, 0.0]
lo = -20
hi = 20

# ---------- Version (i): with continue ----------
total = 0 
count = 0 
dropped = 0

for r in returns:
    if r < lo or r > hi: 
        dropped += 1
        continue

    total += r
    count += 1

mean_1 = total / count        # divide AFTER the loop
print("Version (i):  mean =", mean_1, "| dropped =", dropped)

# ---------- Version (ii): without continue ----------
total = 0 
count = 0
dropped = 0

for r in returns:
    if lo <= r <= hi:
        total += r
        count += 1
    else:
        dropped += 1

mean_2 = total / count
print("Version (ii): mean =", mean_2, "| dropped =", dropped)
```
